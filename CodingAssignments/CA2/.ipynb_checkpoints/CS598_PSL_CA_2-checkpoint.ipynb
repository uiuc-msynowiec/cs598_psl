### Problem 1 [50 points]

In this problem, we are going to use a simulated example to illustrate the behavior of testing and training errors as the number of predictor increases.

(a) [5 points] Generate a data set with p = 20 (independent) covariates, n = 1,000 observations and an associated quantitative response vector generated according to the model

$Y = Xβ + ε$ with

$β = (2, −1, 0, 0.8, 0, −1.5, 1.2, 0, 0, 0.3, −0.7, 0, 1.8, 0, 0.5, 0, −1.2, 0, 0.4, 0)$

The covariates and the ε can be generated using standard normal random variables. (This
is similar to what you did in Coding Assignment # 1).

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error

RANDOM_SEED = 598
n = 1000

rng = np.random.default_rng(RANDOM_SEED)
X = rng.standard_normal((n, 20))
df = pd.DataFrame(X, columns=[f'X{i}' for i in range(1, 21)])

beta = np.array([2, -1, 0, 0.8, 0, -1.5, 1.2, 0, 0, 0.3, -0.7, 0, 1.8, 0, 0.5, 0, -1.2, 0, 0.4, 0])
eps = rng.standard_normal(n)
y = X @ beta + eps

```

(b) [5 points] Randomly split the data set into a training set containing 200 observations and a test set containing 800 observations.

```python
# split data int 200 training and 800 test observations
x_train, x_test, y_train, y_test = train_test_split(df, y, train_size=200, random_state=RANDOM_SEED)
```

(c) [10 points] Perform best subset selection on the training set, and plot the training set MSE associated with the best model of each size. Recall that

$MSE = {\frac 1 n} \sum\limits_{i=1}^n (y_i − \hat{y}_i)^2$

```python
from sklearn.feature_selection import SequentialFeatureSelector
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

X_tr = x_train.values
X_te = x_test.values

best_models = []
for k in range(1, 21):
    if k == 20:
        cols = list(range(X_tr.shape[1]))
    else:
        selector = SequentialFeatureSelector(
            LinearRegression(),
            n_features_to_select=k,
            direction="forward",
            scoring="neg_mean_squared_error",
            n_jobs=-1,
        ).fit(X_tr, y_train)
        # Which features we should use
        cols = selector.get_support(indices=True).tolist()

    # Actually fit a model with the features determined
    model = LinearRegression().fit(X_tr[:, cols], y_train)
    train_mse = mean_squared_error(y_train, model.predict(X_tr[:, cols]))
    test_mse = mean_squared_error(y_test, model.predict(X_te[:, cols]))
    best_models.append((cols, model, train_mse, test_mse))
```

(d) 10 points] Plot the test set MSE associated with the best model of each size.

```python
test_mses = [e[3] for e in best_models]
best_size = np.argmin(test_mses) + 1
best_test_mse = test_mses[best_size - 1]

plt.plot(range(1, 21), test_mses, marker="x")
plt.scatter(best_size, best_test_mse, color="red", zorder=3, label=f"Minimum: {best_size} predictors")
plt.xlabel("Number of predictors")
plt.ylabel("Test MSE")
plt.title("Best Subset Selection - Test MSE")
plt.legend()
plt.show()
```

(e) [5 points] For which model size does the test set MSE take on its minimum value? Comment on your results.

```python
train_mses = [e[2] for e in best_models]
for i, m in enumerate(best_models):
    print(f"For {i + 1} feature(s), train MSE {m[2]} {">=" if m[2] >= m[3] else "<"} test MSE {m[3]}")
display(best_models[best_size - 1])
```

#### Comments on results:

For subsets of 6 features and less, the test set MSE was less than the training set MSE, though the test set MSE was minimized at 11 features. The training set MSE, however, was minimized only when using all available features in the dataset. The MSE was high for both with few features because the model was underfitted. This correlates with how the data was generated because out of the 20 initial random features, only 11 were used to generate y. At the top end where the model used all 20 features, the training set MSE being lower than the test set MSE, while the test set MSE increases past 11, shows that the model was overfitted to the training data. Past the optimal number of features, 11, the training set MSE shows a drastically scaled-down reduction in MSE, showing diminishing gains due to incorporation of noise from features unrelated to y.

(f) [5 points] How does the model at which the test set MSE is minimized compare to the true model used to generate the data? Comment on the coefficient values.

```python
print("Betas used to generate y:")
print(best_models[best_size - 1][0])
print(beta[best_models[best_size - 1][0]])

beta_dist = []

for i, m in enumerate(best_models):
    if i == best_size - 1:
        print("Model with lowest test MSE:")
    print(f"{i + 1} feature(s) used: {best_models[i][0]}")
    print(beta[best_models[i][0]])
    print(best_models[i][1].coef_)

    beta_hat_full = np.zeros(len(beta))
    beta_hat_full[best_models[i][0]] = best_models[i][1].coef_
    beta_diff = beta - beta_hat_full
    
    print(beta_diff)
    print(beta_diff * beta_diff)
    beta_dist.append(float(np.sqrt(np.sum(beta_diff * beta_diff))))
```

#### Comments on MSE:

The resultant $\hat{\beta}$'s closely resembled the ${\beta}$'s used to generate y. Interestingly, as features were added up to 11, only features used to initially generate y were consecutively added, meaning up until the optimal feature set, only meaningful features were selected. This makes sense since the features that correlate with y would have the biggest impact on minimizing the test MSE. Also, while not completely consistent, the features first selected below the optimal 11 features tended to be those with the largest magnitude of beta when generating y. So for a single feature, feature 0 was selected with a beta of 2.0, followed by feature 5 which was added, with a beta of -1.5. This general trend continued, with the last feature added at the 11 feature set being 18 which had one of the smallest non-zero beta magnitudes (0.4).

This exercise highlighted the imporance and impact of relevant features vs the side effect of irrelevant features decreasing performance with the test set.

### Problem 2 [50 points – 10 points each]

We will use the [Parkinsons Telemonitoring data set](https://doi.org/10.24432/C5ZS3N) from the UCI Machine Learning Repository.

The data contain 5,875 biomedical voice measurements from 42 patients with early-stage Parkinson’s disease. The goal is to predict a textitpatient’s disease severity (UPDRS score) from a collection of voice measures and a few demographic variables.

You should use the following variables as predictors:

| Variable | Description |
| ----------- | ----------- |
| `age` | Subject age |
| `sex` | Subject gender (0 = male, 1 = female) |
| `test-time` | Days since recruitment into the trial |
| `Jitter(%), Jitter(Abs)`<br>`Jitter:RAP, Jitter:PPQ5, Jitter:DDP` | Measures of variation<br>in fundamental frequency |
| `Shimmer, Shimmer(dB), Shimmer:APQ3`<br>`Shimmer:APQ5, Shimmer:APQ11, Shimmer:DDA` | Measures of variation in<br>amplitude |
| `NHR, HNR` | Noise-to-tonal component ratios |
| `RPDE` | Nonlinear dynamical complexity measure |
| `DFA` | Signal fractal scaling exponent |
| `PPE` | Nonlinear measure of fundamental frequency variation |

Response: total UPDRS – clinician’s total UPDRS score, linearly interpolated.

Data preparation: Download the data, remove the columns `subject#` and `motor_UPDRS`, and create a random 70 % / 30 % train–test split. Set the random seed to 598.


```python
# first, pip install ucimlrepo

from ucimlrepo import fetch_ucirepo 
import sklearn
import pandas as pd
  
# fetch dataset 
parkinsons_telemonitoring = fetch_ucirepo(id=189) 

# clean dataset
#parkinsons_telemonitoring = 

# data (as pandas dataframes) 
X = parkinsons_telemonitoring.data.features 
y = parkinsons_telemonitoring.data.targets 
  
# metadata 
print(parkinsons_telemonitoring.metadata) 
  
# variable information 
print(parkinsons_telemonitoring.variables) 

# data prep
y = y.drop(columns="motor_UPDRS")
X_train, X_test, y_train, y_test = sklearn.model_selection.train_test_split(X, y, test_size=0.3, random_state=598)

```

(a) Fit a best-subset selection algorithm (up to 8 predictors, excluding the intercept) and report the best model of each size together with its training and test prediction errors (MSE). Keep the output concise.

```python
import numpy
import itertools
from sklearn.metrics import mean_squared_error
from sklearn.linear_model import LinearRegression
from math import inf

best_subsets = []

for k in range(1,9):
    lowest_train_mse_k = inf
    lowest_test_mse_k = inf
    lowest_rss_k = inf
    
    for x in itertools.combinations(X, k):
        
        train_subset = X_train[list(x)]
        test_subset = X_test[list(x)]

        ### TRAINING 
        lr = LinearRegression()
        lr.fit(train_subset, y_train)
        y_pred_train = lr.predict(train_subset)
        
        # current model mse/rss
        train_mse = mean_squared_error(y_train, y_pred_train)
        n = len(y_train)
        rss_k = n * train_mse

        ### TESTING
        y_pred_test = lr.predict(test_subset)
        test_mse = mean_squared_error(y_test, y_pred_test)

        # rss_k minimized on training error
        if(rss_k < lowest_rss_k):
            lowest_rss_k = rss_k
            lowest_train_mse_k = train_mse
            lowest_test_mse_k = test_mse
            best_cols_k = x
            
    # report best model of each size with training and testing MSE
    print("Best model of size", k)
    print("Model columns:", best_cols_k)
    print("Training MSE:", lowest_train_mse_k)
    print("Testing MSE:", lowest_test_mse_k)
    print("\n")
    best_subsets.append([k, lowest_rss_k, best_cols_k, lowest_train_mse_k, lowest_test_mse_k])
```

(b) Using the models from (a), which model is preferred by (i) AIC, (ii) BIC, (iii) Mallows’ $C_p$, and (iv) adjusted $R^2$? For each of these four chosen models report both the training and the test MSE.

```python
from math import log, inf

# Mallow requires sigma squared
lr_full = LinearRegression()
lr_full.fit(X_train, y_train)
y_pred_full = lr_full.predict(X_train)

full_mse = mean_squared_error(y_train, y_pred_full)
rss_full = n * full_mse
p_full = len(X_train.columns)
sigma2_mallow = rss_full / (n - p_full - 1)

# Adjusted-R-Squared requires Total Sum of Squares (TSS)
tss = ((y_train - y_train.mean()) ** 2).sum()["total_UPDRS"]

# Finding minimum and maximum values 
min_aic = inf
min_bic = inf
min_c_p = inf
max_adj_r_squared = -inf

# Search through each model found in 2(a)
for row in best_subsets:

    # save variables
    k = row[0]
    p = row[0]
    rss = row[1]
    
    # AIC
    aic = (n * log(rss / n)) + (2 * p)
    # BIC
    bic = (n * log(rss / n)) + (log(n) * p)
    # Mallow
    c_p = (rss / sigma2_mallow) + (2 * p) - n
    # Adjusted-R-Squared
    sigma2_r_squared = rss / (n - p - 1)
    sigma2_0_r_squared = tss / (n - 1)    
    adj_r_squared = 1 - (sigma2_r_squared / sigma2_0_r_squared)

    if aic < min_aic:
        min_aic = aic
        min_aic_model = row
    if bic < min_bic:
        min_bic = bic
        min_bic_model = row
    if c_p < min_c_p:
        min_c_p = c_p
        min_c_p_model = row
    if adj_r_squared > max_adj_r_squared:
        max_adj_r_squared = adj_r_squared
        max_adj_r_squared_model = row

# report training and test MSE for each best
print("Model preferred by AIC")
print("Model size:", min_aic_model[0]) # k
print("Model columns:", min_aic_model[2]) # cols
print("Training MSE:", min_aic_model[3]) # train MSE
print("Testing MSE: ", min_aic_model[4]) # test MSE
print("\n")

print("Model preferred by BIC")
print("Model size:", min_bic_model[0]) # k
print("Model columns:", min_bic_model[2]) # cols
print("Training MSE:", min_bic_model[3]) # train MSE
print("Testing MSE: ", min_bic_model[4]) # test MSE
print("\n")

print("Model preferred by Mallow C_p")
print("Model size:", min_c_p_model[0]) # k
print("Model columns:", min_c_p_model[2]) # cols
print("Training MSE:", min_c_p_model[3]) # train MSE
print("Testing MSE: ", min_c_p_model[4]) # test MSE
print("\n")

print("Model preferred by Adjusted-R-squared")
print("Model size:", max_adj_r_squared_model[0]) # k
print("Model columns:", max_adj_r_squared_model[2]) # cols
print("Training MSE:", max_adj_r_squared_model[3]) # train MSE
print("Testing MSE: ", max_adj_r_squared_model[4]) # test MSE
```

(c) Fit a ridge-regression model to predict `total_UPDRS`. Use cross-validation to select the regularization parameter λ.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import RidgeCV

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
y_train = y_train.squeeze()

alphas = [1e-3, 1e-2, 1e-1, 1, 10, 100, 1000]
ridge_regression_model = RidgeCV(alphas=alphas).fit(X_train_scaled,y_train)
print("Regularization Parameter:", ridge_regression_model.alpha_)
pred_total_UPDRS_ridge_regression_model = ridge_regression_model.predict(X_test_scaled)
```

(d) Fit a lasso model on the same data. Identify which features are driven to exactly zero.

```python
from sklearn.linear_model import LassoCV

lasso_model = LassoCV(max_iter=100000, alphas=alphas, tol=1e-3).fit(X_train_scaled,y_train)
print("Regularization Parameter:", lasso_model.alpha_)
pred_total_UPDRS_lasso_model = lasso_model.predict(X_test_scaled)

zero_features = X_train.columns[lasso_model.coef_ == 0][0]
print("Features Driven to Exactly Zero:", zero_features)
```

(e) Compare the predictive performance (test MSE) of the models selected in (b) with ridge and with lasso. Which approach would you recommend for predicting total UPDRS and why?

```python
test_mse_ridge_regression_model = mean_squared_error(y_test, pred_total_UPDRS_ridge_regression_model)
test_mse_lasso_model = mean_squared_error(y_test, pred_total_UPDRS_lasso_model)

# ridge never zeroes out coefficients, so its param count is always all 19 predictors
ridge_nonzero_params = int((ridge_regression_model.coef_ != 0).sum())
lasso_nonzero_params = int((lasso_model.coef_ != 0).sum())

print("Test MSE for Ridge-Regression Model:", test_mse_ridge_regression_model, "| # params:", ridge_nonzero_params)
print("Test MSE for Lasso Model:", test_mse_lasso_model, "| # params:", lasso_nonzero_params)

print("Test MSE for Model preferred by AIC:", min_aic_model[4], "| # params:", min_aic_model[0])
print("Test MSE for Model preferred by BIC:", min_bic_model[4], "| # params:", min_bic_model[0])
print("Test MSE for Model preferred by Mallow C_p:", min_c_p_model[4], "| # params:", min_c_p_model[0])
print("Test MSE for Model preferred by Adjusted-R-squared:", max_adj_r_squared_model[4], "| # params:", max_adj_r_squared_model[0])
```

I would recommend the lasso model for predicting total UPDRS because it has the lowest test MSE of all the models under consideration.
